Q8.check score type of all students in the datasets

```python
score_columns = ['Math Score', 'Reading Score', 'Writing Score']
df_scores_melted = data[score_columns].melt(var_name='Score Type', value_name='Score Value')

df_scores_melted = df_scores_melted.dropna(subset=['Score Value'])

plt.figure(figsize=(8, 6))
sns.countplot(x='Score Type', data=df_scores_melted,palette="pastel")

plt.xlabel('Score Type')
plt.ylabel('Number of Students with Score')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

Q13. comparison of Reading Score  Distribution by daily sleep duration

```python
plt.figure(figsize=(12, 7))
sns.kdeplot(data['Study Time per Week'].dropna(), fill=True, color='skyblue', label='Study Time per Week', linewidth=2)
sns.kdeplot(data['Daily Sleep Duration'].dropna(), fill=True, color='lightcoral', label='Daily Sleep Duration', linewidth=2)
plt.title('Comparison of Study Time per Week and Daily Sleep Duration')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend(title='Variable')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

Q15. Plot Continuous Variables in the dataset

```python
cate_val = []
cont_val = []
for column in data.columns:
    print(column)

data[column]

len(data[column].unique())
len(data[column].unique())<=10
```

```python
cate_val = []
cont_val = []
for column in data.columns:
    if data[column].dtype == 'object' or data[column].dtype == 'bool':
        if len(data[column].unique()) <= 10:
            cate_val.append(column)
        else:
            pass
    else:
        cont_val.append(column)

print("Categorical columns:", cate_val)
print("Continuous columns:", cont_val)

plt.figure(figsize=(15,6))
data[cont_val].hist(figsize=(15,10))
plt.tight_layout()
plt.show()
```
